```python
from pathlib import Path
import pandas as pd
from IPython.display import display, Markdown

# notebooks/ 기준 프로젝트 루트
PROJECT_ROOT = Path.cwd().parents[0]

csv_path = PROJECT_ROOT / "reports" / "real_user_eval_sheet.csv"

print("CSV path:", csv_path)
print("exists:", csv_path.exists())

df = pd.read_csv(csv_path)

print("shape:", df.shape)
```

```python
def print_human_eval_row_full(row, idx=None):
    title = f"[{idx}] " if idx is not None else ""

    print("=" * 120)
    print(f"{title}질문")
    print("-" * 120)
    print(row.get("1. 실제 사용자 질문 (Question)", ""))

    print("\nRAG 최종 답변")
    print("-" * 120)
    print(row.get("3. RAG 최종 답변 (Response)", ""))

    print("\nRAG 참고 문서")
    print("-" * 120)
    print(row.get("2. RAG 참고 문서 (Context)", ""))

    print("\n팀원 평가")
    print("-" * 120)

    score_cols = [
        "질문 적합성 점수 (1~5)",
        "핵심 정보 포함 점수 (1~5)",
        "근거 일치성 점수 (1~5)",
        "환각 여부 (Y/N)",
        "실무 활용 가능성 점수 (1~5)",
        "코멘트",
    ]

    for col in score_cols:
        print(f"{col}: {row.get(col, '')}")

    print("=" * 120)
    print()
```

```python
# 특정 행 하나만 전체 출력
print_human_eval_row_full(df.iloc[32], idx=0)
```
